## 1. Alinhamento

**Pergunta delimitada:**

Como a palavra "liberdade" é empregada e ressignificada nos discursos de posse presidencial brasileiros entre 1985 e 2023?

**Finalidade e estrutura analítica:**

Finalidade descritiva e comparativa, com estrutura comparativa entre períodos (1985/2023) e etapa explicativa situada via contexto histórico.

**Unidade de análise:**

O trecho/parágrafo do discurso contendo a palavra "liberdade" e seu contexto sintático e semântico.

**Que observações seriam necessárias para responder:**

A distribuição e frequência das palavras coocorrentes ao termo "liberdade", a categorização das dimensões de seu uso no discurso presidencial (ex.: liberdade civil, política, econômica, de expressão ou religiosa) e a comparação temporal e ideológica da evolução desses contextos semânticos entre as diferentes gestões da Redemocratização (1985–2023).

## 2. Fontes e cadeia de produção

**Fontes primárias em relação à pergunta:**

Transcrições oficiais dos discursos formais de posse dos presidentes brasileiros no Congresso Nacional (1985–2023).

**Fontes secundárias e dados derivados:**

Obras teóricas sobre História dos Conceitos (Koselleck) e Dados Situados (D'Ignazio & Klein), dados derivados compostos por arquivos .txt limpos, tabela de metadados e chunks de texto.

**Instituições, produtores e custodiantes:**

Poder Executivo Federal / Presidência da República (produtores originais); Congresso Nacional, Biblioteca da Presidência e portais de imprensa/documentação pública (custodiantes e distribuidores atuais).

**Finalidade original dos registros e categorias herdadas:**

Pronunciamento ritualístico e institucional de tomada de posse perante a nação, herdou-se a linguagem formal do rito de posse, a estrutura de oratória política executiva e as convenções da comunicação oficial.

**Condições materiais e técnicas de acesso:**

Acesso digital via arquivos de texto corrido nativos em .txt, exigindo apenas limpeza simples de caracteres e normalização de encoding (UTF-8).

## 3. População, corpus e seleção

**População de interesse:**

O discurso político e oficial do Poder Executivo Federal sobre o conceito de "liberdade" durante a transição da ditadura militar para a Redemocratização no Brasil (1985–2023).

**População acessível:**

Transcrições e pronunciamentos oficiais dos discursos de posse dos presidentes brasileiros proferidos perante o Congresso Nacional entre 1985 e 2023, preservados em acervos públicos digitais (Biblioteca da Presidência, Congresso Nacional e portais de documentação pública).

**Corpus previsto:**

Coleção delimitada de discursos presidenciais brasileiros no intervalo 1985–2023 com ocorrências explícitas do termo "liberdade".

| Critério | Campo/evidência | Regra | Justificativa | Caso limítrofe |
|---|---|---|---|---|
| inclusão | `texto_conteudo` / `ano` | Presença literal do termo "liberdade" e ano entre 1985 e 2023 | Garantir pertinência temática e temporal na redemocratização | Variações no plural ("liberdades") são incluídas |
| exclusão | `tipo_discurso` | Discursos proferidos fora da cerimônia oficial de posse | Manter o padrão do gênero discursivo | Discursos de vitória na noite da eleição são excluídos |

**Registro das exclusões:**

Registrado na tabela de descarte (motivos_exclusao.csv), armazenando o ID do documento, tipo de descarte (fora do período, gênero discursivo não compatível) e a devida justificativa.

## 4. Cobertura, vieses e silêncios

| Dimensão | Cobertura | Lacuna | Efeito sobre a análise | Mitigação |
|---|---|---|---|---|
| **temporal** | 1985 a 2023 | Intervalos de 4 anos entre os mandatos e pronunciamentos rituais pontuais | A análise se restringe aos momentos de posse, sem cobrir discursos cotidianos | Mapear o contexto político e eleitoral específico de cada ano de posse |
| **espacial** | Brasília (Congresso Nacional) | Outras esferas de fala pública nos estados e municípios | Visão centralizada na capital federal e no governo nacional | Declarar o escopo do discurso executivo federal central |
| **social** | Elites políticas e governantes | Vozes da sociedade civil, trabalhadores, movimentos sociais e oposição | Predomínio da retórica institucional, apaziguadora e de cúpula | Reler criticamente os silêncios e limitar o alcance da interpretação |
| **institucional** | Presidência da República | Pronunciamentos do Poder Legislativo e do Judiciário | Foco exclusivo na visão do Chefe do Poder Executivo | Contextualizar a composição partidária da base de governo no ano |
| **documental** | Discursos formais de posse proferidos | Entrevistas, planos de governo e cartas pessoais | Análise restrita à modalidade de oratória protocolar da posse | Explicitar que a análise trata estritamente da retórica oficial de posse |

**Possíveis mecanismos de viés de seleção:**

Viés da institucionalidade do gênero discursivo de posse, que tende a adotar tom conciliador, genérico e de construção de consenso, podendo distorcer disputas ideológicas mais agudas.

**Silêncios não redutíveis a valores ausentes:**

A voz de grupos marginalizados ou movimentos sociais que não possuem representação direta na elaboração da oratória oficial da Presidência da República.

**Limites das afirmações:**

As conclusões dirão respeito exclusivamente à evolução conceitual da palavra "liberdade" na oratória formal de posse dos presidentes da Nova República (1985–2023), sem generalização direta para a opinião pública geral ou para a totalidade da sociedade brasileira.

## 5. Metadados, identificadores e proveniência

**Identificador, escopo e persistência:** Identificador alfanumérico único (id_doc, ex.: DISCURSO_2003_LULA) gerado na ingestão do arquivo .txt, persistente na base e independente de alterações nos atributos do arquivo.

| Campo | Definição | Tipo/domínio | Origem | Regra |
|---|---|---|---|---|
| `id_doc` | Identificador do discurso | Texto | Ingestão | DISCURSO_ANO_PRESIDENTE (ex: DISCURSO_2003_LULA) |
| `ano` | Ano da posse | Número (1985–2023) | Cabeçalho do texto | 4 dígitos |
| `presidente` | Nome do presidente | Texto | Ingestão | Nome político padronizado |
| `partido` | Sigla partidária | Categoria | Catalogação | Sigla oficial da posse |
| `texto_conteudo` | Transcrição integral | Texto livre | .txt nativo | Texto corrido limpo em UTF-8 |
| `trecho_id` | Identificador do parágrafo | Texto | Processamento | id_doc + _PAR_N |

**Metadados administrativos e de direitos:**
Documentos de acesso público e transparência oficial do Estado brasileiro (discursos formais de posse presidencial proferidos no Congresso Nacional), livres de restrições de direitos autorais para fins de ensino e pesquisa acadêmica.

**Como a fonte será ligada ao registro derivado:**
Cada trecho/parágrafo extraído preservará em seus metadados o id_doc (ex.: DISCURSO_2003_LULA), o nome do arquivo .txt fonte e o número do parágrafo de origem para garantir auditabilidade.

**Versão, data de acesso, agente e transformações:**
Gerenciados via arquivo de proveniência em JSON (proveniencia_pipeline.json), registrando data de acesso/download dos arquivos .txt, regras de limpeza, normalização de caracteres e parâmetros do script de segmentação de parágrafos.

Exactamente 12 transcrições integrais de discursos em .txt (1985 a 2023), resultando em uma estimativa de 60 a 120 trechos/parágrafos de análise na tabela processável.  

**Amostra piloto:**
Piloto executado com 3 discursos em .txt (José Sarney/1985, FHC/1995 e Lula/2003) para validação da rotina de extração de parágrafos e contagem de coocorrência lexical em Pandas.

**Dependências de autorização ou infraestrutura:**
Acesso à internet para obtenção das transcrições oficiais e ambiente básico Python (Jupyter/Colab com biblioteca Pandas).

**Plano B se a principal fonte não puder ser usada:**

**O que será deliberadamente deixado de fora:**
Discursos cotidianos e de improviso, declarações e entrevistas à imprensa, pronunciamentos em rádio/TV e oratórias proferidas fora do rito oficial de posse.
