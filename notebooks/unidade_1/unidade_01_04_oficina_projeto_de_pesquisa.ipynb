### Minha cadeia de evidência

**Operação prevista:**

Extração dos parágrafos contendo a palavra "liberdade" nas 12 transcrições de posse (1985–2023).

**Resultado possível:**

Alteração no padrão de termos associados à palavra "liberdade", migrando de "democracia/redemocratização" (1985–1995) para "mercado/eficiência" ou "expressão/direitos civis" nos mandatos subsequentes.

**De que esse resultado poderia ser evidência?**

Da redefinição retórica do conceito de liberdade pelo Poder Executivo Federal, acompanhando as mudanças de prioridades políticas e ideológicas dos governos na Nova República.

**O que ele não demonstraria diretamente?**

A efetiva aplicação prática de tais liberdades em políticas públicas ou a percepção da população civil sobre o conceito.  

**Interpretações alternativas:**

A variação lexical pode refletir apenas convenções do estilo pessoal do presidente/redator do discurso ou a retórica conciliadora típica do rito de posse, em vez de uma mudança ideológica profunda.

**Casos ou documentos que precisariam ser relidos:**

Discursos de posse específicos em momentos de forte polarização (ex.: 1990, 2019 e 2023) e pronunciamentos de vitória das mesmas gestões.

**Conhecimento contextual necessário:**
