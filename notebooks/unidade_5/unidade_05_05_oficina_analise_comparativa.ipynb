## 1. Pergunta e contexto

**Pergunta comparativa:** A taxa de uso do termo "liberdade/liberdades" (por
mil palavras) é sistematicamente maior nos discursos de posse proferidos nas
bordas do período democrático de 1985–2023 (1985, 1990, 1995, 2019, 2023) do
que nos proferidos no centro dessa série (1992, 1999, 2003, 2007, 2011, 2015,
2016)?

**Contexto histórico, social, linguístico ou cultural:** O corpus cobre da
redemocratização (posse de Sarney, 1985) à quarta posse de Lula (2023),
atravessando dois processos de impeachment (Collor→Itamar, 1992; Dilma→Temer,
2016) e duas polarizações eleitorais agudas (2018, 2022). "Liberdade" é um
termo cujo uso em discurso de posse tende a se intensificar quando o regime
democrático está sob disputa explícita, e a rarear quando não está.

**Literatura que torna a comparação pertinente:** Koselleck (Futuro Passado)
trata conceitos como "liberdade" como historicamente variáveis, não como
rótulos estáveis, o que justifica perguntar quando, não só quanto, o
termo é mobilizado. D'Ignazio & Klein fundamentam o cuidado com silêncios e
ausências (os três discursos com zero ocorrências) como dado, não como falha
de coleta.

**Por que comparar estes grupos ou documentos:** O agrupamento Pontas × Centro
não é um corte arbitrário de calendário: nasceu da observação, no Notebook 03
de visualização (Unidade 4), de um padrão em forma de U na taxa por discurso, alto nas duas pontas da série, baixo no meio. Comparar os dois grupos testa se
esse padrão visual se sustenta como diferença robusta.

A pergunta delimita o contraste; agora verifique se as unidades e os
subconjuntos são comparáveis e quais assimetrias precisam permanecer visíveis.

## 3. Especificação principal

**Percurso escolhido:** combinado (quantitativo com apoio textual).

**Quantidade, representação ou variável:** `taxa_liberdade_mil` — ocorrências
de "liberdade/liberdades" por 1.000 palavras, por discurso (numérica de
razão). A contagem bruta foi descartada como quantidade principal porque a
extensão dos discursos varia quase 10× (614 a 5.917 palavras, Notebook 01,
seção de sensibilidade).

**Medida principal:** diferença de médias entre Pontas e Centro, acompanhada
do tamanho de efeito padronizado (d de Cohen com desvio combinado) e, para
quantificar a variabilidade sob reamostragem, um intervalo de 95% via
bootstrap e um valor de p via teste de permutação.

**Fórmula ou convenção implementada:** `d = (média_Pontas − média_Centro) /
desvio_combinado`, com `desvio_combinado` pela fórmula pooled-variance (n₁−1 e
n₂−1 como pesos); bootstrap com reposição dentro de cada grupo, B=4000;
permutação dos rótulos Pontas/Centro sobre os 12 valores combinados, M=5000,
p bilateral com correção +1 no numerador e denominador.

**Pré-processamento e parâmetros:** contagem de `liberdade(s)` por regex
`\bliberdades?\b` sobre o texto normalizado (NFC, espaços colapsados);
denominador = tokens do documento via `re.findall(r"[a-záàâãéêíóôõúüçñ]+")`
em minúsculas, sem *stemming*.

**Por que esta medida responde à pergunta:** a pergunta é sobre intensidade de
uso controlada pelo tamanho do texto, não sobre presença/ausência nem sobre
vocabulário geral do discurso, por isso a taxa, e não a contagem bruta nem
uma medida de similaridade lexical inteira (essas ficam para a especificação
alternativa e para os casos, abaixo).

Com a especificação fixada, relate o resultado sem antecipar explicação. Inclua
denominadores, unidades, escala e tabela equivalente a qualquer gráfico.

## 5. Especificação alternativa e sensibilidade

**Alternativa escolhida:** retirada justificada de caso influente (Sarney
1985, a maior taxa do corpus), complementada por uma segunda alternativa,
cosseno TF-IDF de vocabulário completo (Notebook 03/04), para checar se a
proximidade lexical geral entre discursos segue o mesmo agrupamento.

**Razão metodológica:** Sarney sozinho responde por 19 das 55 ocorrências do
corpus; antes de aceitar a diferença Pontas/Centro como padrão, é preciso
verificar se ela sobrevive sem o caso mais extremo.

**Resultado alternativo:** sem Sarney, a diferença Pontas − Centro cai de
2,493 para 1,973 ocorrências/mil palavras (Notebook 01, célula `sec5`). Já a
similaridade de cosseno TF-IDF (Notebook 04) não segue o mesmo agrupamento: o
discurso vetorialmente mais parecido com Lula 2023 (que é "Pontas") é Dilma
2015 (que é "Centro"), com cosseno 0,371, maior inclusive que qualquer par
formado só com discursos de Pontas.

**O que permaneceu e o que mudou:** a diferença de taxa permanece grande e no
mesmo sentido mesmo sem o caso mais influente, não depende de um único
discurso. Mas a similaridade vetorial de vocabulário geral (medida diferente,
sobre o texto inteiro) não reproduz o agrupamento Pontas/Centro: dois
discursos podem compartilhar vocabulário amplo mesmo tendo taxas de
"liberdade" muito diferentes (Notebook 04, seção de inspeção de casos).

**A conclusão depende da escolha?** Depende de qual pergunta se faz. Para "a
taxa do termo-alvo difere entre os grupos", a resposta é robusta às duas
alternativas de sensibilidade quantitativa testadas. Para "os discursos de
cada grupo formam clusters de vocabulário reconhecíveis", a resposta é não,
a similaridade vetorial geral mede outra coisa, e não deve ser lida como
confirmação do padrão de taxa.

A sensibilidade compara agregados. Retorne agora aos registros ou trechos que
sustentam, contradizem ou explicam a divergência entre medidas.

## 7. Interpretação, relevância e limites

**Afirmação diretamente sustentada:** nos 12 discursos de posse presidencial
de 1985 a 2023, a taxa de ocorrência de "liberdade/liberdades" por mil
palavras é maior nos discursos de 1985, 1990, 1995, 2019 e 2023 do que nos de
1992, 1999, 2003, 2007, 2011, 2015 e 2016, e essa diferença não desaparece
quando o caso mais extremo (Sarney) é removido.

**Relevância substantiva da diferença ou semelhança:** com médias de 2,842 e
0,349 e quase nenhuma sobreposição entre as faixas de valores dos dois grupos,
a diferença é grande em termos absolutos, não só estatisticamente destacável, o que já havia sido registrado como critério no Notebook 00 desta unidade
("em um corpus completo, a diferença descritiva pode ser o resultado
relevante").

**Interpretação contextual:** o padrão é consistente com a leitura de que o
termo é mobilizado em momentos de disputa aberta sobre o regime democrático
(a própria redemocratização em 1985; a polarização de 2018–2023), e menos
presente quando a legitimidade democrática não está em xeque no mesmo grau.

**Hipóteses alternativas:** (a) efeito de gênero retórico, discursos mais
curtos e programáticos (Sarney) podem concentrar vocabulário de forma
diferente de discursos longos; (b) efeito de redator/assessoria de discurso,
não do presidente ou da conjuntura; (c) a fronteira Pontas/Centro foi definida
depois de observar o padrão em U (H2, Unidade 4), é uma hipótese pós-hoc, e
outra divisão razoável dos 12 anos mudaria o resultado.

**O que não pode ser concluído:** que há uma tendência histórica contínua
(a série liga anos não consecutivos); que o valor de p = 0,0018 mede a
probabilidade de a hipótese nula ser verdadeira (Notebook 00 desta unidade);
que o uso do termo reflete convicção pessoal do presidente, e não seleção de
discurso, conjuntura ou redação.

**Limites de seleção, medição, representação e generalização:** n=12,
população completa e não amostra, nada aqui generaliza para discursos
futuros. A contagem depende da regex `\bliberdades?\b` (exclui `livre`,
`livremente`); a segmentação por sentença substituiu parágrafos ausentes em
11 dos 12 arquivos; `partido` está integralmente ausente da base
(`razao_partido_ausente`); e, como o caso de Lula 2023 mostra, taxa alta pode
significar afirmação ou recusa do conceito, a medida quantitativa não
resolve essa ambiguidade sozinha.

medida, casos e conclusão.

## 10. Síntese e entrega

**Resumo da pergunta, método e resultado:** perguntei se a taxa de
"liberdade" por mil palavras é maior nas bordas do período 1985–2023 do que no
centro. Medi a diferença de médias entre os grupos Pontas (n=5) e Centro
(n=7), com d padronizado, IC bootstrap e teste de permutação. Resultado:
diferença de 2,493 ocorrências/mil palavras, d=2,930, IC 95% [1,608; 3,587],
p=0,0018.

**Como a especificação alternativa alterou ou confirmou a leitura:** a
retirada de Sarney (o caso mais influente) reduziu a diferença para 1,973,
mas não a eliminou, o padrão não depende de um único discurso. Já a
alternativa por cosseno TF-IDF de vocabulário geral não confirmou o mesmo
agrupamento, o que limita a interpretação: o padrão vale para o termo-alvo,
não para a semelhança lexical dos discursos como um todo.

**Como os casos qualificaram o agregado:** o caso típico (Bolsonaro 2019)
confirmou o padrão sem exceção. O caso influente (Sarney 1985) mostrou que
parte do pico é formulação aforística, não argumentação distribuída. O caso
que desafiou a leitura (Lula 2023) revelou que taxa alta pode significar
recusa, não afirmação, do conceito, a diferença de médias por si só não
captura essa polaridade.

**Conclusão substantiva limitada:** nestes 12 discursos completos (não numa
amostra), a taxa de uso do termo "liberdade" é consistentemente maior nas
bordas do período democrático de 1985–2023 do que no seu centro, e essa
diferença é robusta à retirada do caso mais extremo, mas o que o termo
significa em cada ocorrência (afirmação ou disputa do conceito) varia dentro
do próprio grupo de taxa alta, e só aparece na leitura de trecho, não na
contagem.
